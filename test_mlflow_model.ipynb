```python
import mlflow
import pandas as pd
from sklearn.metrics import classification_report

mlflow.set_tracking_uri("sqlite:///mlflow.db")
data = pd.read_csv("./data/production_data.csv")
features = ["Air temperature [K]", "Process temperature [K]", "Rotational speed [rpm]", "Torque [Nm]",
                "Tool wear [min]", "Type_L", "Type_M", "Type_H"]
y = data["Machine failure"]
X = data[features]
model = mlflow.sklearn.load_model("models:/ai4i-failure@champion")
proba = model.predict_proba(X)[:, 1]
pred = (proba >= 0.4).astype(int)
report = classification_report(y, pred)
print(report)
```
