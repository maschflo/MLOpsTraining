```python
import mlflow
import pandas as pd
from mlflow import MlflowClient
from sklearn.metrics import classification_report

mlflow.set_tracking_uri("sqlite:///mlflow.db")
data = pd.read_csv("./data/production_data.csv")
mv = MlflowClient().get_model_version_by_alias("ai4i-failure", "champion")
threshold = float(mv.tags.get("recommended_threshold", 0.5))
model = mlflow.sklearn.load_model("models:/ai4i-failure@champion")
y = data["Machine failure"]
proba = model.predict_proba(data)[:, 1]
pred = (proba >= threshold).astype(int)
report = classification_report(y, pred)
print(report)
```
