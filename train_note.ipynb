```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

original_path = "./data/ai4i2020.csv"
original = pd.read_csv(original_path)
original.describe()
```

Remove unnecessary columns and one-hot-encode the "Type"-column before creating new CSVs to avoid doing it every time

```python
original.drop(columns=["UDI", "Product ID", "TWF", "HDF", "PWF", "OSF", "RNF"], inplace=True)
original = pd.get_dummies(original, columns=["Type"])
```

```python
production = original.sample(frac=0.2, random_state=22)
production.describe()
```

```python
test_data = original.drop(production.index)
test_data.describe()
```

```python
production.to_csv("./data/production_data.csv", index=False)
test_data.to_csv("./data/test_data.csv", index=False)
```
